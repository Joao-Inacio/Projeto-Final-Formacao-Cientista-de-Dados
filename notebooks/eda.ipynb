```python
df_limpo = df.copy()
df_limpo['ResidenciaDesde'] = df_limpo['ResidenciaDesde'].fillna(df_limpo['ResidenciaDesde'].mode()[0])
df_limpo['historico'] = df_limpo['historico'].fillna(df_limpo['historico'].mode()[0])
df_limpo['emprego'] = df_limpo['emprego'].fillna(df_limpo['emprego'].mode()[0])
df_limpo['tipohabitacao'] = df_limpo['tipohabitacao'].fillna(df_limpo['tipohabitacao'].mode()[0])
df_limpo['nomeprofissao'] = df_limpo['nomeprofissao'].fillna(df_limpo['nomeprofissao'].mode()[0])
df_limpo.isnull().sum()
```

```python
df_limpo['ResidenciaDesde'] = df_limpo['ResidenciaDesde'].astype(int)
df_limpo
```

```python
df_limpo.describe()
```

```python
variaveis = ['Valor', 'Duracao', 'TempoParcelamento', 'ResidenciaDesde',
    'Idade', 'EmprestimoExistente', 'Dependentes', 'SocioEmpresa',
    'Estrangeiro',]
for variavel in variaveis:
    sns.boxplot(data=df_limpo, y=df_limpo[variavel]);
    plt.title(variavel)
    plt.show()
```

```python
variaveis = ['Valor', 'Duracao', 'Idade']
for variavel in variaveis:
    valor_chao_caixa = df_limpo[variavel].quantile(0.25)
    valor_teto_caixa = df_limpo[variavel].quantile(0.75)

    altura_da_caixa = valor_teto_caixa - valor_chao_caixa
    distancia_segura = altura_da_caixa * 1.5

    limite_inferior = valor_chao_caixa - distancia_segura
    limite_superior = valor_teto_caixa + distancia_segura

    df_limpo[variavel] = df_limpo[variavel].clip(limite_inferior, limite_superior)

    sns.boxplot(data=df_limpo, x=df_limpo[variavel]);
    plt.title(variavel)
    plt.show()
```

```python
sns.histplot(data=df_limpo, x=df_limpo['Status']);
```

```python
variaveis_categoricas = ['historico', 'proposito', 'investimento',
    'emprego', 'EstadoCivil', 'FiadorTerceiro', 'outrosfinanciamentos',
    'tipohabitacao', 'nomeprofissao']
for variavel in variaveis_categoricas:
    plt.figure(figsize=(15, 10))
    sns.countplot(data=df_limpo, x=df_limpo[variavel], hue=df_limpo['Status']);
    plt.title(variavel)
    plt.show()
```
